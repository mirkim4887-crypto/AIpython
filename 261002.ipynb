```python
import random

# 1: 가위, 2: 바위, 3: 보
com = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
    print('컴퓨터: 가위')
elif com == 2:
    print('컴퓨터: 바위')
else:
    print('컴퓨터: 보')

if user == 1:
    print('유저: 가위')
elif user == 2:
    print('유저: 바위')
else:
    print('유저: 보')

if user == com:
    print('서로 비겼습니다.')
elif user == 1 and com == 3:
    print('유저가 이겼습니다!')        # 가위 vs 보
elif user == 2 and com == 1:
    print('유저가 이겼습니다!')        # 바위 vs 가위
elif user == 3 and com == 2:
    print('유저가 이겼습니다!')        # 보 vs 바위
else:
    print('유저가 졌습니다.')
```

```python
def asmd(x,y,op):
    if op == '+':
        print(f'{x} + {y} = {x+y}')
    elif op == '-':
        print(f'{x} - {y} = {x-y}')
    elif op == '+':
        print(f'{x} X {y} = {x*y}')
    elif op == '/':
        print(f'{x} % {y} = {x/y}')
    else:
        print(f'연산자가 잘못되었습니다.')
```

```python
x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op = input('연산자를 입력하시오')
asmd(x,y,op)
```
