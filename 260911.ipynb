```markdown
### 파이썬 `print()` 함수 배우기

`print()` 함수는 데이터를 화면에 출력할 때 사용합니다.
* 문자열을 출력할 때는 따옴표(`' '` 또는 `" "`)를 사용합니다.
* 숫자는 따옴표 없이 입력합니다.
* 여러 값을 출력할 때는 쉼표(`,`)를 사용하며, 기본적으로 값 사이에 공백이 추가됩니다.
```

````python
# 1. 문자열 출력
print("안녕하세요! 파이썬 공부를 시작합니다.")

# 2. 숫자 출력
print(2024)

# 3. 여러 값 출력 (쉼표 사용)
print("올해는", 2024, "년입니다.")

# 4. 연산 결과 출력
print(100 + 200)
````
