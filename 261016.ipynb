```python
#1부터 100가지 합을 출력하시오
sum = 0
for a in range (1,11):
  sum += a
print(sum)
```

```python
#1,100까지 짝수 합
#if문 안쓰기
s = 0
for a in range (2,101,2):
  s += a
print(s)

#if문 쓰기
s = 0
for a in range (101):
  if a%2 == 0:
    s += a
  else:
    s += 0
print(s)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요
a = int(input('구구단을 출력할 정수를 입력하시오: '))
for x in range(1,10):
  print(f'{a} X {x} = {a*x}')
```

```python
#2단부터 9단까지
for i in range(2,10):
  for j in range (1,10):
    print(f'{i} X {j} = {i*j}')
  print('-------------------')
```

```python
#while문으로 1~100까지 합
a = 0
i = 1
while i<=100:
  a+=i
  i+=1
print(a)
```

```python
#while문으로 1~100까지 합
a = 0
i = 1
while True:
  a+=i
  i+=1
  if i>100:
    break
print(a)
```

```python
# 양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요.
i=0
while True:
  a = int(input('정수를 입력하시오'))
  if a == 0:
    print(f'합계는 {i}입니다')
    break
  i += a
```

```python
scores = [85, 95, 70, 75, 99, 93, 88]

print('최댓값:', max(scores))
print('최솟값:', min(scores))
print('합:', sum(scores))
print('평균:', sum(scores)/len(scores))
```
