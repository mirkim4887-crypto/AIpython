```python
# ch02_plus3_order1.py
americano_num = int(input('아메리카노 주문 수: '))
latte_num = int(input('라떼 주문 수: '))
green_tea_num = int(input('녹차 주문 수: '))

sum = americano_num * 2000 + latte_num * 3000 + green_tea_num * 2500

print(f'전체 주문 금액은 {sum:,}원입니다.')
```
